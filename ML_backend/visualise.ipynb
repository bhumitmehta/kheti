```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Load the data
data = pd.read_csv('data.csv')

# Scatter Plot: Nitrogen (N) vs Potassium (K)
plt.figure(figsize=(8, 6))
plt.scatter(data['N'], data['K'], alpha=0.7, c='blue', edgecolors='w')
plt.title("Nitrogen vs Potassium")
plt.xlabel("Nitrogen (N)")
plt.ylabel("Potassium (K)")
plt.grid()
plt.show()

# Histogram: Temperature distribution
plt.figure(figsize=(8, 6))
plt.hist(data['temperature'], bins=15, color='skyblue', edgecolor='black')
plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")
plt.grid()
plt.show()

# Bar Plot: Average N, P, K by crop type
avg_npk = data.groupby('label')[['N', 'P', 'K']].mean()
avg_npk.plot(kind='bar', figsize=(10, 6))
plt.title("Average N, P, K by Crop Type")
plt.xlabel("Crop Type")
plt.ylabel("Average N, P, K")
plt.grid()
plt.show()

# Scatter Plot: Temperature vs Humidity colored by crop type

# Get unique crop types and assign a color to each one
crop_types = data['label'].unique()
colors = plt.cm.tab20(np.linspace(0, 1, len(crop_types))) # Use np.linspace to get different colors

# Plot Temperature vs Humidity with different colors for each crop type
plt.figure(figsize=(10, 6))

for i, crop in enumerate(crop_types):
    subset = data[data['label'] == crop]
    plt.scatter(subset['temperature'], subset['humidity'], alpha=0.7, label=crop, color=colors[i])

plt.title("Temperature vs Humidity by Crop Type")
plt.xlabel("Temperature")
plt.ylabel("Humidity")
plt.legend(title="Crop Type")
plt.grid()
plt.show()

# Boxplot: pH values by crop type
plt.figure(figsize=(10, 6))
data.boxplot(column='ph', by='label', grid=False)
plt.title("pH Levels by Crop Type")
plt.suptitle("")
plt.xlabel("Crop Type")
plt.ylabel("pH")
plt.xticks(rotation=90)  # Rotate x-axis labels to vertical
plt.show()

```
